## 1. 한글 폰트 설정 (맨 처음 한 번만)인터넷에서 나눔고딕 폰트를 받아 matplotlib에 등록합니다.

```python
from pyodide.http import pyfetch
import matplotlib.font_manager as fm
import matplotlib.pyplot as plt

FONT_URL = "https://cdn.jsdelivr.net/gh/google/fonts@main/ofl/nanumgothic/NanumGothic-Regular.ttf"

res = await pyfetch(FONT_URL)                 # 폰트 파일 내려받기
with open("NanumGothic.ttf", "wb") as f:
    f.write(await res.bytes())

fm.fontManager.addfont("NanumGothic.ttf")     # matplotlib에 폰트 등록
plt.rcParams["font.family"] = "NanumGothic"
plt.rcParams["axes.unicode_minus"] = False    # 마이너스 기호가 깨지는 것 방지

print("한글 폰트 준비 완료")
```

## 2. 잘 되는지 확인아래를 실행했을 때 제목과 축 이름의 한글이 제대로 보이면 성공입니다.

```python
import matplotlib.pyplot as plt

동 = ["고잔동", "초지동", "선부동", "와동"]
온도 = [31.2, 30.4, 32.1, 29.8]

plt.figure(figsize=(5, 3))
plt.bar(동, 온도)
plt.title("행정동별 여름철 평균 지표 온도")
plt.xlabel("행정동")
plt.ylabel("온도(℃)")
plt.show()
```
